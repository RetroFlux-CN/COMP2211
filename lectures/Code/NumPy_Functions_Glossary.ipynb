#### [array](https://numpy.org/doc/stable/reference/generated/numpy.array.html)

The most basic function creates a NumPy array from a list, tuple, or another NumPy array. Returns a fresh, new array.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
print(a1)

a2 = np.array([[1, 5, 7], [2, 6, 4]])
print(a2)
```

#### [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)

Changes the shape of a NumPy array. When one of the arguments is -1, then NumPy will automatically figure out the dimension for you. Returns a copy of the new array.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
# Reshapes to (3,2)
b1 = a1.reshape(3,2)
print(b1)

# This also reshapes the array to (3,2), but NumPy figured out the 3 for you
c1 = a1.reshape(-1,2)
print(c1)

# Another example
a2 = np.array([[1, 5, 7], [2, 6, 4]])
b2 = a2.reshape(6)
print(b2)

# Question: what is the output of a1.reshape(1, -1) and a1.reshape(-1, 1)?
```

#### [Transpose](https://numpy.org/doc/stable/reference/generated/numpy.transpose.html)

Permutes the axes of a given array and returns the new array. The optional arguments specify the order of the axes to permute the array.

```python
# Creates an array of shape (24,) filled with entries from 0 to 23 in ascending order
range24 = np.arange(24)

# Reshape the range array into a 6 x 4 grid
a1 = range24.reshape(6, 4)
print(a1)

# When the matrix is 2D, then transpose (without additional arguments) swaps the order of the two axes - like matrix transpose
print(a1.transpose())

# Reshape the (24,) array above into shape (2, 3, 4)
a2 = range24.reshape(2, 3, 4)
print(a2.shape)

# When the matrix is more than 2D, then transpose(a, b, c) gives a new array where its axis 0 is the original axis a, its axis 1 is the original axis b and so on
b2 = a2.transpose(1, 2, 0)
print(b2.shape)

# When the matrix is more than 2D, then transpose (without additional arguments) reverses the order of the axes
c2 = a2.transpose()
print(c2.shape)

# Creates an array filled with 1s with shape (22, 11, 20, 12)
a3 = np.ones((22, 11, 20, 12))
b3 = a3.transpose(1, 3, 0, 2)
# Question: what is the shape of b3?
```

#### np.random

`np.random` is a module that allows one to make NumPy arrays with random entries. The most commonly used functions from the `np.random` module is `random` and `randint`

#### [min](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.min.html), [max](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.max.html)

Gets the minimum (or maximum) value of an array. If you specify the "axis" argument, the return value will be a new array with $n-1$ dimension, where we take the minimum (or maximum) value along the specified axis only.

We will discuss about the axis argument below.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
# The maximum value in the array is 7, so the output is 7
print(np.max(a1))

a2 = np.array([[1, 5, 7], [2, 6, 4]])
# Refer to the above - we count axes inside out - so (1, 2) is axis 0, and (1, 5, 7) is axis 1
# Take minimum of (1, 2), (5, 6) and (7, 4) and return a NumPy array with these three elements
print(np.min(a2, axis = 0))

# Question: make an array of shape (2,) that consists of the largest element of each row from a2
```

#### [The `axis` argument](https://stackoverflow.com/questions/22320534/how-does-the-axis-parameter-from-numpy-work)

When we specify the `axis` argument in an array method, we only apply the method across a specified axis. The returned array has the same shape except that the specified dimension will be missing. If we specify `axis = n`, then we perform the operation along the axis corresponding to `some_aray.shape[n]`.

For instance, if the array `some_array` has shape `(1, 2, 3, 4, 5, 6)` and we perform `np.min(some_array, axis = 3)`, then the result has shape `(1, 2, 3, 5, 6)`.

Intuitively, we "squash" the array along the chosen axis, and perform whatever operations on the numbers that get squashed together.

```python
# Initialize an array with random entries between 0, 1 and shape (22, 11, 12, 3, 4)
a1 = np.random.random((22, 11, 12, 34, 56))

# Get the minimum values along the 2nd axis
b1 = np.min(a1, axis = 2)
c1 = np.min(b1, axis = 3)

# Question: what is the shape of b1 and c1?

# Uses broadcasting to tile an array
a2 = np.arange(15) + np.zeros((5, 10, 25, 1), dtype = int)

# Question: Describe the array a2. What is its shape? What are its entries?
```

#### [argmin](https://numpy.org/doc/stable/reference/generated/numpy.argmin.html), [argmax](https://numpy.org/doc/stable/reference/generated/numpy.argmax.html)

Get the index of the minimum (or maximum) value of an array. If you specify the "axis" argument, the return value will be a new array with $n-1$ dimension, where we take the minimum (or maximum) value along the specified axis only.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
# The index of the maximum value (7) is 2, so the output is 2
print(a1.argmax())

a2 = np.array([[1, 5, 7], [2, 6, 4]])
# Question: what does np.argmin(a2) return?

# Take minimum of (1, 2), (5, 6) and (7, 4) and return a NumPy array with the indices (along the specified axis) of these three elements
# Index of 1 is 0, index of 5 is 0, and index of 4 is 1, so the output is [0, 0, 1]
print(a2.argmin(axis = 0))
```

#### [argsort](https://numpy.org/doc/stable/reference/generated/numpy.argsort.html)

In argsort, the first element of the array will be the index of the minimum value of the array; the second element of the array will be the index of the second minimum value, and so on. This has an optional `axis` argument.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
# Minimum value (1) is at index 0
# Second minimum value (2) is at index 3
# Third minimum value (4) is at index 5
# ...
print(np.argsort(a1))

a2 = np.array([[1, 5, 7], [2, 6, 4]])
# The axis argument does exactly the same thing as above
print(np.argsort(a2, axis = 0))

# Question: How do we get the index of the second largest element of a1?

# Question: What should a1[np.argsort(a1)] return?
```

#### [all](https://numpy.org/doc/stable/reference/generated/numpy.all.html)
Returns True if all elements in a given array is True, and False otherwise. If the elements in an array are numbers, then returns True if all elements are non-zero, and False otherwise.

This function is very useful for array comparison since we can't directly use `==` on two NumPy arrays (This returns another NumPy array)

This has an optional `axis` argument.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
a2 = np.array([1, 5, 7, 2, 6, 4])
a3 = np.array([1, 5, 7, 2, 6, 3])
a4 = np.array([0, 4, 6, 1, 5, 3])

# Question: what does a1 == a2 return?

print(np.all(a1 == a2))
print(np.all(a1 == a3))
print(np.all(a1 != a4))
```

#### [count_nonzero](https://numpy.org/doc/stable/reference/generated/numpy.count_nonzero.html)
Returns an integer which counts the number of `True`s in a NumPy array. If the elements in an array are numbers, then returns an integer that counts the number of non-zero elements

This function is very useful for array comparison since we can't directly use `==` on two NumPy arrays (This returns another NumPy array). In addition, your TA likes this function a bit more since it gives you more control over `np.all`

This has an optional `axis` argument.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
a2 = np.array([1, 5, 7, 2, 6, 4])
a3 = np.array([1, 5, 7, 2, 6, 3])
a4 = np.array([0, 4, 6, 1, 5, 3])

print(np.count_nonzero(a1 == a3))

# If we want to perform the "array equal" comparison between a1 and a2 like above, we can do the following:
# We can instead look at a1 != a2, then these two arrays are equal if there are no Trues in a1 != a2
print(np.count_nonzero(a1 != a2) == 0)
print(np.count_nonzero(a1 != a3) == 0)
```

### Inter-array operations

#### [multiply](https://numpy.org/doc/stable/reference/generated/numpy.multiply.html)

Perform element-wise multiplication between two arrays. For NumPy arrays, this function demonstrates the same behaviour as `a1 * a2`

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
a2 = np.array([3, 1, 1, 2, 8, 7])

# Performs element-wise multiplication on arrays
print(a1 * a2)
print(np.multiply(a1, a2))
```

#### [dot](https://numpy.org/doc/stable/reference/generated/numpy.dot.html)

This function essentially performs matrix multiplication between two arrays.

If `a1` and `a2` are both 1D arrays, then `np.dot(a1, a2)` returns the dot product of the vectors `a1` and `a2` (element-wise multiply then sum over all entries)

If `a1` and `a2` are both 2D arrays, then `np.dot(a1, a2)` returns the matrix product of `a1` and `a2`.

If `a1` and `a2` does not satisfy one of the above two cases, then NumPy performs some overloads for the function; you can read about its behaviour [here](https://numpy.org/doc/stable/reference/generated/numpy.dot.html).

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
a2 = np.array([3, 1, 1, 2, 8, 7])

print(np.dot(a1, a2))
# Question: write some code to confirm that the vector dot product is really "element-wise multiply then sum over all entries".

a3 = np.array([[-1, -5,  0], [ 0, -3,  1], [ 0,  2, -1]])
a4 = np.array([[-2,  5,  0], [ 1,  0,  4], [-4,  3,  0]])
print(np.dot(a3, a4))
```

#### [concatenate](https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html)
Merges two or more NumPy arrays along a specific axis. The dimensions of the NumPy arrays must match, except possibly at the specified axis.

The arrays should be provided to the function as a single tuple.

```python
a1 = np.array([1, 5, 7, 2, 6, 4])
a2 = np.array([6, 7, 8, 11])


# Joins a1 and a2, head to tail
a3 = np.concatenate((a1, a2))
print(a3)


# We provide another demonstration of concatenate, on arrays with higher dimensions
b1 = np.array([[1, 5, 7], [2, 6, 4]])
b2 = np.array([[6, 7], [8, 11]])

# Joins a1 and a2 head to tail along axis 1 (the innermost axis)
b3 = np.concatenate((b1, b2), axis = 1)
print(b3)
```
